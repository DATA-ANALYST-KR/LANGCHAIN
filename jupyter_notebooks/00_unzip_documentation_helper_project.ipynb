# Unzip the Documentation Helper Project

Use this notebook to extract:

```text
documentation_helper_gemini_streamlit.zip
```

into the folder containing the course notebooks.

After extraction, the folder structure will be:

```text
documentation_helper_gemini/
├── backend/
│   ├── __init__.py
│   └── core.py
└── main.py
```


## 1. Imports

````python
from pathlib import Path
from zipfile import ZipFile
````

## 2. Define the ZIP file and extraction location

Place this notebook and the ZIP file in the same folder before running the cell.


````python
CURRENT_FOLDER = Path.cwd()

ZIP_FILE = (
    CURRENT_FOLDER
    / "documentation_helper_gemini_streamlit.zip"
)

EXTRACT_FOLDER = CURRENT_FOLDER


print("Current folder:", CURRENT_FOLDER)
print("ZIP file:", ZIP_FILE)
print("Extraction folder:", EXTRACT_FOLDER)
````

## 3. Extract the project

Running this cell extracts the files into the current notebook folder.

````python
with ZipFile(ZIP_FILE, "r") as zip_file:
    zip_file.extractall(EXTRACT_FOLDER)


print("Project extracted successfully.")
````

## 4. Display the extracted files

````python
PROJECT_FOLDER = (
    EXTRACT_FOLDER
    / "documentation_helper_gemini"
)


for file_path in PROJECT_FOLDER.rglob("*"):
    if file_path.is_file():
        print(
            file_path.relative_to(
                EXTRACT_FOLDER
            )
        )
````

## 5. Expected output

```text
documentation_helper_gemini/main.py
documentation_helper_gemini/backend/__init__.py
documentation_helper_gemini/backend/core.py
```

To launch the application, open an Anaconda Prompt, activate `nlp_env`, move into the extracted project folder, and run:
