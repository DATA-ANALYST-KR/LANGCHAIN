### PINECONE INDEX CHECKS

```python
import os
from pinecone import Pinecone


pinecone_client = Pinecone(
    api_key=os.environ[
        "PINECONE_API_KEY"
    ]
)


index_name = os.environ[
    "DOCUMENTATION_INDEX_NAME"
]


index_description = (
    pinecone_client.describe_index(
        index_name
    )
)


print(
    "Index name:",
    index_description.name,
)

print(
    "Dimension:",
    index_description.dimension,
)

print(
    "Metric:",
    index_description.metric,
)

print(
    "Ready:",
    index_description.status["ready"],
)

print(
    "Host:",
    index_description.host,
)
```
